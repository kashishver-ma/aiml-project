```python
#expand short forms 

#it has a built in dictonary

chat_word_dict = {
    "u": "you",
    "r": "are",
    "b4": "before",
    "gr8": "great",
    "l8r": "later",
    "idk": "I don’t know",
    "btw": "by the way",
    "imo": "in my opinion",
    "ttyl": "talk to you later",
    "brb": "be right back",
    "wanna": "want to",
    "gonna": "going to",
    "lemme": "let me",
    "dunno": "don’t know",
    "thx": "thanks",
    "pls": "please",
    "ppl": "people",
    "smh": "shaking my head",
    "tbh": "to be honest",
    "omg": "oh my god",
    "lol": "laugh out loud",
    "rofl": "rolling on the floor laughing",
    "afk": "away from keyboard",
    "np": "no problem",
    "gtg": "got to go",
    "jk": "just kidding",
    "hbu": "how about you",
    "ily": "I love you",
    "ttys": "talk to you soon"
}

print(chat_word_dict)

text="omg idk what r u talking about lol"

def normalise_text(text):
    words=text.split()  # 
    normalised=[chat_word_dict.get(word.lower(),word) for word in words] #get key = give value
    return " ".join(normalised)

normalise_text(text=text)
```
