```python
#n grams of nlp

from nltk.util import ngrams
from nltk.tokenize import word_tokenize

text="i am kashish verma"
tokens=word_tokenize(text)
n=2;

bigrams=list(ngrams(tokens,n))
bigrams
```

```python
# tfidf

from sklearn.feature_extraction.text import TfidfVectorizer

text=["i love nlp","i love ml","i love dl"]
vectoriser=TfidfVectorizer()
tfidf_matrix=vectoriser.fit_transform(text) #upto this
print(tfidf_matrix.toarray())

#to represent  it beautifully
feature_name=vectoriser.get_feature_names_out()
# (i love ... are features)

for doc_index,doc in enumerate(tfidf_matrix.toarray()):
    #emuerate used to break into 2 parts
    print("document",doc_index+1)

    for word_index,score in enumerate(doc):
        if score>0:
            print(f"{feature_name[word_index]} : {score:.4f}")
    
                  



```
