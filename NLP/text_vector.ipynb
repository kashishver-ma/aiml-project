```python
import spacy

from spacy.cli import download

download("en_core_web_sm")

#nlpo = object of speciallym library spacy
nlp=spacy.load('en_core_web_sm') #built in in old models (old model name)
# nlp=spacy.load('')

text="hello,#Welcome 123@ to nNLP"
#we need to normal this text

#function to clean with argument
def clean_text(text):


    cleaned_text="".join(char for char in text if char.isalpha() or char.isspace())
    doc=nlp(cleaned_text)  #nlp return it in form of documents

    return "".join(token.text for token in doc) #token is word

clean=clean_text(text=text)
print(clean)

```

```python
# nltk = text pre-processing or vectorisation library

import nltk

nltk.download('punkt_tab')
# nltk.download('punkt')

text="hello, #welcome 123 @to NLP"
tokens=nltk.word_tokenize(text) #convert to token
print(tokens)

clean_token=[token for token in tokens if token.isalnum()] #alphanumeric
clear="".join(clean_token)
print("clear" ,clear)
```
