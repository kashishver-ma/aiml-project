```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
```

```python
IMG_ROWS, IMG_COLS = 28, 28
NUM_CLASSES = 10
```

```python
x_train = x_train.reshape(x_train.shape[0], IMG_ROWS, IMG_COLS, 1)
x_test = x_test.reshape(x_test.shape[0], IMG_ROWS, IMG_COLS, 1)
input_shape = (IMG_ROWS, IMG_COLS, 1)
```

```python
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255
```

```python
print(f"x_train shape: {x_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"Number of training samples: {x_train.shape[0]}")
print(f"Number of test samples: {x_test.shape[0]}")
```

```python
y_train = to_categorical(y_train, NUM_CLASSES)
y_test = to_categorical(y_test, NUM_CLASSES)
```

```python
model = Sequential([
    # First Convolutional Block
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=input_shape),
    MaxPooling2D(pool_size=(2, 2)),

    # Second Convolutional Block
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),
    
    Dropout(0.25), # Regularization to prevent overfitting

    # Fully Connected Layers
    Flatten(), # Prepare output for the Dense layers
    Dense(128, activation='relu'),
    Dropout(0.5), # More regularization

    # Output Layer: 10 units for 10 classes, using Softmax for probability
    Dense(NUM_CLASSES, activation='softmax')
])
```

```python
model.compile(
    loss='categorical_crossentropy', # Appropriate loss for multi-class classification
    optimizer='adam', # A good default optimizer
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
batch_size = 128
epochs = 10

print("\nStarting model training...")
history = model.fit(
    x_train, y_train,
    batch_size=batch_size,
    epochs=epochs,
    verbose=1,
    validation_data=(x_test, y_test)
)
```

```python
print("\nEvaluating model performance...")
score = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Loss: {score[0]:.4f}")
print(f"Test Accuracy: {score[1]*100:.2f}%")
```
