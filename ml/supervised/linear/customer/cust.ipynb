```python
import pandas as pd
df = pd.read_csv('cust_satisfaction.csv')
df
```

```python
df.drop_duplicates
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.drop(['Age','Flight Distance','Gender'],inplace=True, axis = 1)
```

```python
df
```

```python
print(df['Type of Travel'].unique())
df['satisfaction'].unique()
df['Class'].unique()
```

```python
dct = {
  'Loyal Customer':1, 'disloyal Customer':0
}
dct1 = {
  'Personal Travel':1, 'Business travel':0
}
dct2= {
  'satisfied':1, 'neutral or dissatisfied':0
}

dict3 = {
  'Eco':2, 'Eco Plus':0, 'Business':1}

print(dct2)
print(dict3)
```

```python
df.dropna(inplace=True)
df['Customer Type']=df['Customer Type'].map(dct)
df['Type of Travel']=df['Type of Travel'].map(dct1)
df['satisfaction']=df['satisfaction'].map(dct2)
df['Class']=df['Class'].map(dict3)
df.columns
```

```python
df.head()

x=df.drop('Customer Type', axis=1)

y=df['Customer Type']

#we have to tell customer type-loyal disloyal

print(df)
# df.columns

```

```python

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42)
x_train.shape, x_test.shape, y_train.shape, y_test.shape

x_train.columns

```

```python
#appplying model - classification

from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(x_train, y_train)


lr.score(x_train, y_train)
lr.score(x_test, y_test)
lr.predict(x_test)


import joblib
#object nm , "model.lb" -------> saved model
model = joblib.dump(lr, "custmodel.lb") 

model = joblib.load("custmodel.lb")
predlr=model.predict(x_test)
print(predlr)

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(x_train, y_train)
rf.score(x_train, y_train)
rf.score(x_test, y_test)
rf.predict(x_test)

import joblib
#object nm , "model.lb" -------> saved model
model = joblib.dump(rf, "custmodel1.lb")
model = joblib.load("custmodel1.lb")
pred=model.predict(x_test)

pred

for(i,j) in zip(pred, y_test):
    print(i,j)









```
